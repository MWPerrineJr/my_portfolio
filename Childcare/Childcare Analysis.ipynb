<h1><center>Visualizations</center></h1>

```python
# Bar Chart built with altair comparing median household income for Texas residents.
chart_b = alt.Chart(txInc).mark_bar(
    stroke='black',
    strokeWidth=1
).encode(
    x = alt.X('MHI', title= 'Income'),
    y = alt.Y('TotalPop', title='Population'),
    tooltip=['MHI', 'TotalPop']
).properties(
    title = 'Median HouseHold Income in Texas',
    width = 600,
    height = 400
    
).interactive()
chart_b
```

```python
# Bar Chart built with altair comparing median household income for US residents.
chart_b_us = alt.Chart(ccInc).mark_bar(
    stroke='black',
    strokeWidth=1
).encode(
    x = alt.X('MHI', title= 'Income'),
    y = alt.Y('TotalPop', title='Population'),
    tooltip=['MHI', 'TotalPop']
).properties(
    title = 'Median US HouseHold Income',
    width = 600,
    height = 400
    
).interactive()
chart_b_us
```

```python
# Bar chart built with Altair to compare average Texas childcare costs.
chart_h = alt.Chart(texasCC).mark_bar().encode(
    x=alt.X('Date:T', timeUnit='year', bin=True, 
            title = 'Year'),
    y = alt.Y('Ave_Cost:Q', title = 'Texas CC Costs')
   
).properties(
    title = 'Annual Average Cost of Child Care in Texas',
    width = 600,
    height = 400
).interactive()
chart_h
```

```python
# Bar chart built with Altair to compare average US childcare costs. 
ccCost_chartS = alt.Chart(texasCC).mark_bar().encode(
    x=alt.X('Date:T', timeUnit='year', bin=True, 
            title = 'Year'),
    y = alt.Y('Ave_Cost:Q', title = 'Average Cost of Childcare')
   
).properties(
    title = 'Annual National Average Cost of Child Care',
    width = 600,
    height = 400
).interactive()
ccCost_chartS
```

```python
# Build scatter plot with altair to compare US median income by year.
alt.data_transformers.disable_max_rows()
ccInc_chartS = alt.Chart(ccInc).mark_point().encode(
    alt.X('Date:T', timeUnit='year', axis=alt.Axis(ticks=False),
          title = 'Date'),
    alt.Y('MHI:Q', title = 'National Median Household Income'),
    color = 'State_Abbreviation',
    shape = 'State_Abbreviation'
).properties(
    title = 'National Household Income',
    width = 600,
    height= 600
).interactive()
ccInc_chartS
```

```python
# Build scatter plot with altair to compare Texas median income by year.
txInc_chartS = alt.Chart(txInc).mark_circle().encode(
    alt.X('Date:T', timeUnit='year', axis=alt.Axis(ticks=False),
          title = 'Date'),
    alt.Y('MHI:Q', title = 'Median Income in Texas'),
    color = 'County_Name',
    shape = 'County_Name'
).properties(
    title = 'Median Texas Income by County',
    width = 600,
    height = 600
).interactive()
txInc_chartS
```

```python
# Create a seaborn boxplot to compare Texas population.
plt.figure(figsize=(10,10))
sns.boxplot(txInc, x = 'Date', y = 'TotalPop')
plt.xticks(rotation=90)
plt.title('Total Population in Texas', fontsize = 20, weight ='bold')
plt.xlabel('Year', fontsize = 15, weight = 'bold')
plt.ylabel('Total Texas Populatiion', fontsize = 15, weight = 'bold')
plt.show()
```

```python
# Create a seaborn boxplot to compare US population.
plt.figure(figsize=(10,10))
sns.boxplot(ccInc, x = 'Date', y = 'TotalPop')
plt.xticks(rotation=90)
plt.title('Total Population in the US', fontsize = 20, weight ='bold')
plt.xlabel('Year', fontsize = 15, weight = 'bold')
plt.ylabel('Total US Populatiion', fontsize = 15, weight = 'bold')
plt.show()
```

```python
# Create a scatter plot with seaborn comparing male and female median income in Texas.
plt.figure(figsize=(10,10))
custom = [Line2D([], [], marker='.', color='r', linestyle='None'),
          Line2D([], [], marker='.', color='b', linestyle='None')]
sns.scatterplot(txInc, x = 'FME', y = 'MME', hue= 'Date', palette = ['red', 'blue'],
                alpha = 0.5)
plt.title('Male Median Income VS Female Median Income in Texas', 
          fontsize = 20, weight = 'bold')
plt.xlabel('Female Median Income', fontsize = 15, weight = 'bold')
plt.ylabel('Male Median Income', fontsize = 15, weight = 'bold')
plt.legend(custom, ['FME', 'MME'], loc='upper right')

plt.show()
```

```python
# Create a scatter plot with seaborn comparing male and female median income in US.
plt.figure(figsize=(10,10))
custom = [Line2D([], [], marker='.', color='r', linestyle='None'),
          Line2D([], [], marker='.', color='b', linestyle='None')]
sns.scatterplot(ccInc, x = 'FME', y = 'MME', hue= 'Date', palette = ['red', 'blue'], 
                alpha = 0.5)
plt.title('Male Median Income VS Female Median Income in US', 
          fontsize = 20, weight = 'bold')
plt.xlabel('Female Median Income', fontsize = 15, weight = 'bold')
plt.ylabel('Male Median Income', fontsize = 15, weight = 'bold')
plt.legend(custom, ['FME', 'MME'], loc='upper right')

plt.show()
```
